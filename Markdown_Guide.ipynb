```
import pandas as pd
import matplotlib.pyplot as plt

# =====================================================
# DADOS CONSOLIDADOS DA WEG S.A. - 2017 A 2021
# Valores em R$ milhões
# =====================================================

dados_balanco = {
    'Conta': [
        'Ativo Circulante',
        'Ativo Não Circulante',
        'Ativo Total',
        'Passivo Circulante',
        'Passivo Não Circulante',
        'Patrimônio Líquido'
    ],

    '2017': [
        9415.667,
        4570.320,
        13985.987,
        4326.788,
        2815.892,
        6843.307
    ],

    '2018': [
        9438.581,
        5961.269,
        15399.850,
        5034.004,
        2512.589,
        7853.257
    ],

    '2019': [
        9760.902,
        5926.739,
        15687.641,
        4491.021,
        2266.630,
        8929.990
    ],

    '2020': [
        12556.143,
        7371.753,
        19927.896,
        5882.044,
        2115.554,
        11930.298
    ],

    '2021': [
        15945.946,
        7986.841,
        23932.787,
        7927.884,
        1994.231,
        14010.672
    ]
}

df_balanco = pd.DataFrame(dados_balanco)

# =====================================================
# DRE - RECEITA E LUCRO LÍQUIDO
# =====================================================

dados_dre = {
    'Conta': [
        'Receita Líquida',
        'Lucro Líquido'
    ],

    '2017': [
        9523.8,
        1142.15
    ],

    '2018': [
        11970.090,
        1338.32
    ],

    '2019': [
        13347.434,
        1614.58
    ],

    '2020': [
        17469.557,
        2340.87
    ],

    '2021': [
        23563.338,
        3585.95
    ]
}

df_dre = pd.DataFrame(dados_dre)

print("Dados carregados com sucesso!")

print("\nBalanço Patrimonial:")
display(df_balanco)

print("\nDRE:")
display(df_dre)
```

```
# ============================================
# PASSO 3 - ANÁLISE VERTICAL E HORIZONTAL
# ============================================

# ---------- ANÁLISE VERTICAL ----------

# DRE: Receita Líquida de 2021 = base 100%
receita_2021 = df_dre.loc[
    df_dre['Conta'] == 'Receita Líquida', '2021'
].values[0]

df_dre['AV 2021 (%)'] = (
    df_dre['2021'] / receita_2021
) * 100


# Balanço: Ativo Total de 2021 = base 100%
ativo_total_2021 = df_balanco.loc[
    df_balanco['Conta'] == 'Ativo Total', '2021'
].values[0]

df_balanco['AV 2021 (%)'] = (
    df_balanco['2021'] / ativo_total_2021
) * 100


# ---------- ANÁLISE HORIZONTAL ----------

# 2017 será o ano-base
df_dre['AH 2017-2021 (%)'] = (
    (df_dre['2021'] - df_dre['2017'])
    / df_dre['2017']
) * 100

df_balanco['AH 2017-2021 (%)'] = (
    (df_balanco['2021'] - df_balanco['2017'])
    / df_balanco['2017']
) * 100


# ---------- EXIBINDO OS RESULTADOS ----------

print("ANÁLISE VERTICAL E HORIZONTAL - BALANÇO PATRIMONIAL")
display(df_balanco)

print("ANÁLISE VERTICAL E HORIZONTAL - DRE")
display(df_dre)
```

```
# ============================================
# PASSO 4 - INDICADORES FINANCEIROS
# ============================================

anos = ['2017', '2018', '2019', '2020', '2021']

# Pegando os valores necessários do Balanço
ac = df_balanco.loc[
    df_balanco['Conta'] == 'Ativo Circulante',
    anos
].values[0]

pc = df_balanco.loc[
    df_balanco['Conta'] == 'Passivo Circulante',
    anos
].values[0]

pnc = df_balanco.loc[
    df_balanco['Conta'] == 'Passivo Não Circulante',
    anos
].values[0]

pl = df_balanco.loc[
    df_balanco['Conta'] == 'Patrimônio Líquido',
    anos
].values[0]

# Pegando o Lucro Líquido da DRE
lucro = df_dre.loc[
    df_dre['Conta'] == 'Lucro Líquido',
    anos
].values[0]


# 1. LIQUIDEZ CORRENTE
# Ativo Circulante / Passivo Circulante

liquidez_corrente = ac / pc


# 2. PARTICIPAÇÃO DE CAPITAL DE TERCEIROS
# (Passivo Circulante + Passivo Não Circulante) / Patrimônio Líquido

capital_terceiros = (pc + pnc) / pl


# 3. ROE - RETORNO SOBRE O PATRIMÔNIO LÍQUIDO
# Lucro Líquido / Patrimônio Líquido

roe = (lucro / pl) * 100


# Criando o DataFrame dos indicadores

df_indicadores = pd.DataFrame({
    'Ano': anos,
    'Liquidez Corrente': liquidez_corrente,
    'Capital de Terceiros': capital_terceiros,
    'ROE (%)': roe
})


# Arredondando os resultados

df_indicadores['Liquidez Corrente'] = (
    df_indicadores['Liquidez Corrente'].round(2)
)

df_indicadores['Capital de Terceiros'] = (
    df_indicadores['Capital de Terceiros'].round(2)
)

df_indicadores['ROE (%)'] = (
    df_indicadores['ROE (%)'].round(2)
)


print("INDICADORES FINANCEIROS - WEG S.A.")
display(df_indicadores)
```

```
# ============================================
# PASSO 5 - GRÁFICOS DOS INDICADORES
# ============================================

# Gráfico 1 - Liquidez Corrente

plt.figure(figsize=(8, 5))

plt.plot(
    df_indicadores['Ano'],
    df_indicadores['Liquidez Corrente'],
    marker='o',
    linewidth=2
)

plt.title('Evolução da Liquidez Corrente - WEG S.A.')
plt.xlabel('Ano')
plt.ylabel('Liquidez Corrente')
plt.grid(True)

plt.show()


# Gráfico 2 - Capital de Terceiros

plt.figure(figsize=(8, 5))

plt.plot(
    df_indicadores['Ano'],
    df_indicadores['Capital de Terceiros'],
    marker='o',
    linewidth=2
)

plt.title('Evolução da Participação de Capital de Terceiros - WEG S.A.')
plt.xlabel('Ano')
plt.ylabel('Capital de Terceiros')
plt.grid(True)

plt.show()


# Gráfico 3 - ROE

plt.figure(figsize=(8, 5))

plt.plot(
    df_indicadores['Ano'],
    df_indicadores['ROE (%)'],
    marker='o',
    linewidth=2
)

plt.title('Evolução do ROE - WEG S.A.')
plt.xlabel('Ano')
plt.ylabel('ROE (%)')
plt.grid(True)

plt.show()
```

## 5. Conclusão

A análise econômico-financeira da WEG S.A. no período de 2017 a 2021 apresentou uma evolução positiva dos principais indicadores analisados.

Na análise de liquidez, a Liquidez Corrente permaneceu acima de 1 em todos os anos, indicando que os ativos circulantes foram superiores aos passivos circulantes durante todo o período analisado.

Em relação à estrutura de capital, a participação de capital de terceiros apresentou redução ao longo do período, passando de 1,04 em 2017 para 0,71 em 2021. Esse resultado demonstra uma menor dependência relativa de capital de terceiros.

O ROE apresentou crescimento de 16,69% em 2017 para 25,59% em 2021, indicando aumento da rentabilidade do patrimônio líquido no período.

Também foi observada uma evolução significativa do lucro líquido, que passou de **R$ 1.142,15 milhões em 2017 para R$ 3.585,95 milhões em 2021**. Dessa forma, os resultados analisados indicam uma evolução positiva da situação econômico-financeira da WEG S.A. no período estudado.
